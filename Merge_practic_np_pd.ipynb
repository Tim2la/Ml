```python
def get_level(x):
    if pd.isna(x):
        return "unknown"
    elif x >= 85:
        return "excellent"
    elif x >= 70:
        return "good"
    else:
        return "bad"
```

```python
full_df["level"] = full_df["score"].apply(get_level)
full_df
```

```python
full_df["hours_squared"] = full_df["hours"] ** 2
full_df
```

```python
import pandas as pd
import numpy as np

employees = pd.DataFrame({
    "employee_id": [1, 2, 3, 4, 5, 6],
    "name": ["Alex", "Max", "Anna", "Kate", "John", "Lisa"],
    "department": ["IT", "IT", "Sales", "Sales", "IT", "Sales"]
})

performance = pd.DataFrame({
    "employee_id": [1, 2, 3, 4, 5],
    "tasks": [15, 8, 20, 11, 18],
    "rating": [4.8, 3.2, 4.5, 2.9, 4.1]
})
```

```python
full_df = employees.merge(performance, on="employee_id", how = "left")
full_df
```

```python
full_df.groupby("department")["rating"].agg(["mean", "min", "max"])
```

```python
full_df.groupby("department").agg({
    "rating": ["mean", "max"],
    "tasks":["mean", "sum"]
})
```

```python
def get_status(rating):
    if pd.isna(rating):
        return "unknown"
    elif rating >= 4.5:
        return "excellent"
    elif rating >= 3.5:
        return "good"
    else:
        return "low"

full_df["status"] = full_df["rating"].apply(get_status)
full_df
```

```python
full_df["tasks_double"] = full_df["tasks"] * 2
full_df
```
