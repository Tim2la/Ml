```python
import pandas as pd
import numpy as np

students = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 108, 109],
    "name": [
        "Alex", "Max", "Anna", "Kate", "John",
        "Mike", "Lisa", "Tom", "Tom", "Emma"
    ],
    "age": [
        19, 21, 18, 20, np.nan,
        23, 19, 24, 24, 22
    ],
    "city": [
        "Moscow", "Berlin", "Moscow", "Paris", "Berlin",
        "Moscow", "Paris", "Berlin", "Berlin", "Moscow"
    ],
    "hours": [
        4, 6, 7, 3, 5,
        np.nan, 8, 2, 2, 6
    ]
})

results = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 108, 109],
    "score": [72, 85, 91, 58, np.nan, 76, 49, 88],
    "passed": [1, 1, 1, 0, 0, 1, 0, 1]
})

courses = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 109],
    "course": [
        "ML", "ML", "ML", "Python", "Python",
        "ML", "Python", "Python", "ML"
    ]
})
```

```python
students.isna().sum()
```

```python
students.duplicated().sum()
```

```python
students.iloc[students.duplicated()]
```

```python
students_clean = students.copy()[~students.duplicated()]
students_clean
```

```python
students_clean["age"] = students_clean["age"].fillna(students_clean["age"].median())
students_clean
```

```python
students_clean["hours"] = students["hours"].fillna(students_clean["hours"].median()).copy()
students_clean
```

```python
students_clean_new_data_1 = students_clean.iloc[:4, :3]
students_clean_new_data_2 = students_clean.loc[students_clean.index[2:6],["name", "age", "city"]].copy()
```

```python
students_clean_new_data_2
```

```python
mask = students_clean["age"] < 21
students_new_data_young = students_clean[mask].copy()
```

```python
mask_2 = (students_clean["city"] == "Moscow") | (students_clean["city"] == "Paris")
students_new_data_city = students_clean[mask_2].copy()
```

```python
mask_3 = (students_clean["city"] == "Moscow") & (students_clean["hours"] >= 6)
students_clean_new_data_3 = students_clean[mask_3].copy()
```

```python
mask_4 = students_clean["hours"] >= 6
students_clean_new_data_4 = students_clean.loc[mask_4, ["name", "city", "hours"]].copy()
```

```python
full_df = students_clean.merge(results, on="student_id", how="left").merge(courses, on="student_id", how="left")
full_df
```

```python
full_df = pd.get_dummies(
    full_df,
    columns=["city", "course"],
    dtype=int
)
full_df
```

```python
full_df["passed"] = full_df["passed"].astype("boolean")
full_df
```

```python
full_df["score_per_hour"] = full_df["score"] / full_df["hours"]
```
