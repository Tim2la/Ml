***Оба класса распределены равномерно***


```python
df = pd.DataFrame({
    "age": [18, 19, 20, 21, 22, 20, 19, 23, 21, 250],
    "score": [55, 60, 63, 67, 72, 69, 61, 75, 70, 80]
})

df.describe()
```

```python
plt.boxplot(df["age"])
plt.boxplot(df["score"])
plt.grid()
```

```python
age_quantile_1 = df["age"].quantile(0.25)
age_quantile_3 = df["age"].quantile(0.75)

score_quantile_1 = df["score"].quantile(0.25)
score_quantile_3 = df["score"].quantile(0.75)

IQR_age =  age_quantile_3 - age_quantile_1
IQR_score = score_quantile_3 - score_quantile_1

lower_age = age_quantile_1 - 1.5 * IQR_age
upper_age = age_quantile_3 + 1.5 * IQR_age

lower_score = score_quantile_1 - 1.5 * IQR_score
upper_score = score_quantile_3 + 1.5 * IQR_score
```

```python
mask_age = (df["age"] > upper_age) | (df["age"] < lower_age)
df.loc[mask_age, "age"]
```
