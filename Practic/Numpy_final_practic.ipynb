```python
import numpy as np

X = np.array([
    [18, 45, 2],
    [21, 80, 5],
    [19, 65, 4],
    [23, 90, 7],
    [20, 55, 3],
    [25, 95, 8],
    [22, 70, 6],
    [18, 40, 1]
], dtype=float)

y = np.array([0, 1, 1, 1, 0, 1, 1, 0])

age = X[:,0]
exam_score = X[:,1]
study_hours = X[:,2]

samples = X.shape[0]
features = X.shape[1]

print("samples:", samples)
print("features:", features)
```

```python
mean_feature = X.mean(axis=0)
std_feature = X.std(axis=0)

argmax_exam_score = exam_score.argmax()

first_data_puple = X[exam_score >= 70]
second_data_puple = X[(age >= 21) & (study_hours >= 5)]
third_data_puple = exam_score[y == 1]
```

```python
X_scaled = (X - mean_feature)/(std_feature)

print(X_scaled.mean(axis = 0))
print(X_scaled.std(axis = 0))
```

```python
w = np.array([0.4, 1.2, -0.7])
b = -0.3

z = X_scaled @ w +b
p = 1 / (1 + np.exp(-z))

print(p)
```

```python
y_pred = (p >= 0.5).astype(int)

result = (y == y_pred).astype(int).sum()

accuracy = result / len(y)
```
